```python
#THIS CELL CREATES AN EXAMPLE GRAPH OF ERROR ACCUMULATING USING EULER'S METHOD

import matplotlib.pyplot as plt
import numpy as np

# Define the differential equation: dy/dt = f(t, y)
def f(t, y):
    return -y

# Analytical exact solution: y(t) = e^(-t)
def exact_solution(t):
    return np.exp(-t)

# Parameters
t0, tf = 0.0, 5.0
y0 = 1.0
h = 0.25  # Step size

# Time grid
t = np.arange(t0, tf + h, h)
n = len(t)

# Euler's method computation
y_euler = np.zeros(n)
y_euler[0] = y0

for i in range(n - 1):
    y_euler[i + 1] = y_euler[i] + h * f(t[i], y_euler[i])

# Calculate exact values
y_exact = exact_solution(t)

# Plotting the results on a single axis
plt.figure(figsize=(8, 5))
plt.plot(t, y_exact, 'g-', label='Exact Solution ($e^{-t}$)', linewidth=2)
plt.plot(
    t, y_euler, 'bo--', label=f"Euler's Method ($h={h}$)", markersize=5
)

# Graph configurations
plt.xlabel('Time (t)')
plt.ylabel('y(t)')
plt.title("Euler's Method vs. Exact Solution")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
