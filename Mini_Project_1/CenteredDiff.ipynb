Make a table of the error of the three-point centered-difference 
formula for f (0), where f (x) = sinx − cos x, with h = 10^−1,...,10^−12, 
as in the table in Section 5.1.2. Draw a plot of the results. Does the 
minimum error correspond to the theoretical expectation?

```python
# THIS CELL EVALUATES THE DERIVATIVE OF f(x) = sin(x) - cos(x) AT x = 0 USING THE THREE-POINT CENTERED DIFFERENCE FORMULA.
# THE CELL THEN OUTPUTS THE RESULTS IN A TABLE.

import numpy as np

# 1. Define the function
def f(x):
    return np.sin(x) - np.cos(x)

# 2. Define the exact analytical derivative for comparison
# f'(x) = cos(x) + sin(x)
def df_exact(x):
    return np.cos(x) + np.sin(x)

# 3. Setup parameters
x0 = 0.0
exact_val = df_exact(x0)

# Create a range of successively smaller h values (10^-1 down to 10^-12)
h_values = [10**(-i) for i in range(1, 13)]

# 4. Print headers for the output table
print("=== TABLE 1: STANDARD DECIMAL PRECISION ===")
print(f"{'h':<12} | {'Approximation':<18} | {'Absolute Error':<18}")
print("-" * 54)

# 5. Calculate approximations and errors
for h in h_values:
    # Three-point centered difference formula
    approx = (f(x0 + h) - f(x0 - h)) / (2 * h)
    error = abs(approx - exact_val)
    
    print(f"{h:<12.1e} | {approx:<18.10f} | {error:<18.10e}")

# 6. Create a table of results
data = []
for h in h_values:
    approx = (f(x0 + h) - f(x0 - h)) / (2 * h)
    error = abs(approx - exact_val)
    data.append((h, approx, error))
print("\n=== TABLE 2: HIGHER PRECISION (out to 12 decimal places)===")
print(f"{'h':<12} | {'Approximation':<18} | {'Absolute Error':<18}")
print("-" * 54)
for row in data:
    print(f"{row[0]:.1e}{' '*5} | {row[1]:.12f}{' '*4} | {row[2]:.12e}")
```

```python
import numpy as np

# 1. Define the function and its exact analytical derivative
def f(x):
    return np.sin(x) - np.cos(x)

def df_exact(x):
    return np.cos(x) + np.sin(x)

# 2. Setup parameters
x0 = 0.0
exact_val = df_exact(x0)
h_values = [10**(-i) for i in range(1, 13)]

# 3. Print table headers
print(f"{'h':<8} | {'Forward Error':<15} | {'Backward Error':<15} | {'Centered Error':<15}")
print("-" * 60)

# 4. Calculate and compare errors
for h in h_values:
    # Finite difference formulas
    approx_forward  = (f(x0 + h) - f(x0)) / h
    approx_backward = (f(x0) - f(x0 - h)) / h
    approx_centered = (f(x0 + h) - f(x0 - h)) / (2 * h)
    
    # Absolute errors
    err_forward  = abs(approx_forward - exact_val)
    err_backward = abs(approx_backward - exact_val)
    err_centered = abs(approx_centered - exact_val)
    
    # Print results in scientific notation
    print(f"{h:<8.0e} | {err_forward:<15.5e} | {err_backward:<15.5e} | {err_centered:<15.5e}")

```
