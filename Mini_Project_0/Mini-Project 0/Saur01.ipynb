```python
import numpy as np
from HornersMethod import *
```

```python
# --- Computer Problem 1---
d1 = 50
x1 = 1.00001
c1 = np.ones(d1 + 1)

p1_computed = nest(d1, c1, x1)
q1_exact = (x1**(d1 + 1) - 1) / (x1 - 1)
error1 = abs(p1_computed - q1_exact)
relerror1 = abs((q1_exact-p1_computed)/q1_exact)*100

print("--- Problem 1 Results ---")
print(f"Computed P(x): {p1_computed:.15f}")
print(f"Exact Q(x):    {q1_exact:.15f}")
print(f"Absolute Error: {error1:.1e}")
print(f"Relative Error: {relerror1:.1e}%\n")
```

```python
# --- Computer Problem 2 ---
d2 = 99
x2 = 1.00001
c2 = np.array([(-1)**i for i in range(d2 + 1)])

p2_computed = nest(d2, c2, x2)
q2_exact = (1 - x2**(d2 + 1)) / (1 + x2)
error2 = abs(p2_computed - q2_exact)
relerror2 = abs((q2_exact - p2_computed)/q2_exact)*100

print("--- Problem 2 Results ---")
print(f"Computed P(x): {p2_computed:.15f}")
print(f"Exact Q(x):    {q2_exact:.15f}")
print(f"Absolute Error: {error2:.1e}")
print(f"Relative Error: {relerror2:.1e}%")
```
